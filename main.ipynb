# Deteksi Objek Real-time dengan YOLO (yolov3-tiny) + OpenCV

Notebook ini menyediakan langkah cepat untuk:
- Menginstall dependensi
- Mendownload file YOLOv3-tiny (cfg, weights) dan `coco.names` jika belum ada
- Menjalankan skrip `detect.py` untuk deteksi real-time menggunakan webcam


```python
# Install dependencies (jalankan satu kali).
# Jika sudah terpasang, Anda bisa melewati cell ini.
import sys
import subprocess
packages = [
    'opencv-python',
    'numpy'
]
for p in packages:
    subprocess.run([sys.executable, '-m', 'pip', 'install', p], check=True)
print('Dependencies installed (or already present).')
```

```python
# Download YOLOv3-tiny weights/cfg and coco.names into ./yolo if missing
import os
from urllib.request import urlretrieve
os.makedirs('yolo', exist_ok=True)
files = {
    'yolov3-tiny.weights': 'https://pjreddie.com/media/files/yolov3-tiny.weights',
    'yolov3-tiny.cfg': 'https://raw.githubusercontent.com/pjreddie/darknet/master/cfg/yolov3-tiny.cfg',
    'coco.names': 'https://raw.githubusercontent.com/pjreddie/darknet/master/data/coco.names'
}
for name, url in files.items():
    dest = os.path.join('yolo', name)
    if not os.path.exists(dest):
        print(f'Downloading {name}...')
        try:
            urlretrieve(url, dest)
            print('Saved to', dest)
        except Exception as e:
            print('Failed to download', name, e)
    else:
        print(name, 'already exists')
```

## Cara menjalankan

1. Pastikan kamera tersambung dan tidak sedang digunakan aplikasi lain (mis. Zoom).

```python
# Contoh: jalankan detect.py langsung dari notebook (opsional)
# Ini akan membuka jendela OpenCV sehingga jalankan hanya jika environment lokal mendukung GUI.
import subprocess, sys, os
script = os.path.join(os.getcwd(), 'detect.py')
if os.path.exists(script):
    print('Running detect.py... (tekan q pada jendela untuk keluar)')
    subprocess.run([sys.executable, script])
else:
    print('detect.py tidak ditemukan di workspace. Jalankan `python detect.py` di terminal setelah menambahkan file.')
```
